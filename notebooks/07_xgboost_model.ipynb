```python
import pandas as pd
from src.utils.paths import TRANSFORMED_DATA_DIR

df = pd.read_parquet(TRANSFORMED_DATA_DIR / "tabular_data.parquet")
df
```

```python
from datetime import datetime
from src.pipelines.training.data_split import train_test_split

X_train, y_train, X_test, y_test = train_test_split(
    df,
    cutoff_date=datetime(2025, 6, 1, 0, 0, 0),
    target_column_name="target_rides_next_hour"
)

print(f"X_train shape: {X_train.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
past_rides_columns = [col for col in X_train.columns if col.startswith("rides_")]
X_train_only_numeric = X_train[past_rides_columns]
```

```python
import xgboost as xgb

model = xgb.XGBRegressor()
model.fit(X_train_only_numeric, y_train)
```

```python
X_test_only_numeric = X_test[past_rides_columns]
predictions = model.predict(X_test_only_numeric)
predictions
```

```python
from sklearn.metrics import mean_absolute_error

test_mae = mean_absolute_error(y_test, predictions)
print(f"Test MAE: {test_mae=:.4f}")
```

```python
from lazypredict.Supervised import LazyRegressor
from sklearn.metrics import mean_absolute_error

reg = LazyRegressor(
    verbose=0,
    ignore_warnings=False,
    custom_metric=mean_absolute_error,
    random_state=42,
)
models, _ = reg.fit(
    X_train_only_numeric, X_test_only_numeric, y_train, y_test
)

# Lower MAE means better predictions of hourly ride demand.
models = models.rename(columns={"mean_absolute_error": "MAE"})
models = models.sort_values("MAE", ascending=True)
models
```
